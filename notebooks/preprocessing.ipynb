```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully")
```

```python
# Load dataset
df = pd.read_csv("../dataset/credit_risk_dataset.csv")

print("Dataset loaded successfully")
print("Dataset shape:", df.shape)
```

```python
df.info()
```

```python
# Check duplicates before removal
duplicate_count = df.duplicated().sum()

print("Duplicates before removal:", duplicate_count)
```

```python
# Remove duplicates
df = df.drop_duplicates()

print("Duplicates removed")
print("New dataset shape:", df.shape)
```

```python
#to handle missing values first, identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=['int64','float64']).columns
categorical_columns = df.select_dtypes(include=['object']).columns

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
# Handle missing values correctly

# Numerical columns -> Median
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns -> Mode
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])


print("Missing values handled successfully")
```

```python
# Check missing values after handling
missing_after = df.isnull().sum()

missing_after[missing_after > 0]
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicates after removal:", duplicate_count)
```

```python
#to encode Categorical Variables,first Check categorical columns
categorical_columns
```

```python
# Encode categorical variables
label_encoder = LabelEncoder()

for col in categorical_columns:
    df[col] = label_encoder.fit_transform(df[col])

print("Categorical variables encoded successfully")
```

```python
#verrify
df.head()
```

```python
# Separate features and target variable(input and output)

X = df.drop('loan_status', axis=1)
y = df['loan_status']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Feature scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Feature scaling completed")
```

```python
# Convert scaled data back to dataframe
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```
